```python
import pandas as pd
import seaborn as sns
import matplotlib.pyplot as plt
```

```python
df = pd.read_csv('london.csv')
```

```python
df.dtypes
```

Q. 1) Convert the Datatype of 'Date' column to Date-Time format.


```python
df.date = pd.to_datetime(df.date)
```

```python
df
```

Q. 2) Add a new column ''year'' in the dataframe, which contains years only.


```python
df['year'] = df.date.dt.year
```

```python
df
```

(2.B) Add a new column ''month'' as 2nd column in the dataframe, which contains month only.


```python
df.insert(2 , 'month', df.date.dt.month)
```

```python
df
```

Q. 3) Remove the columns 'year' and 'month' from the dataframe.


```python
df.drop(['month', 'year'] , axis=1 , inplace = True)
```

```python
df
```

Data visualization

```python
plt.figure(figsize=(10,5))
sns.lineplot(data=df, x='date', y='average_price', hue='area')
plt.title("Average House Price Trend Over Time by Area")
plt.xlabel("Date")
plt.ylabel("Average Price")
plt.xticks(rotation=45)
plt.show()
```

```python
plt.figure(figsize=(10,5))
sns.lineplot(data=df, x='date', y='houses_sold', hue='area')
plt.title("Number of Houses Sold Over Time by Area")
plt.xlabel("Date")
plt.ylabel("Houses Sold")
plt.xticks(rotation=45)
plt.show()
```

```python
plt.figure(figsize=(8,5))
sns.barplot(data=df, x='area', y='average_price', ci=None)
plt.title("Average House Price by Area")
plt.xlabel("Area")
plt.ylabel("Average Price")
plt.xticks(rotation=45)
plt.show()
```

```python
plt.figure(figsize=(8,5))
sns.scatterplot(data=df, x='no_of_crimes', y='average_price', hue='area')
plt.title("Crimes vs Average House Price")
plt.xlabel("Number of Crimes")
plt.ylabel("Average Price")
plt.show()
```

```python
plt.figure(figsize=(6,4))
sns.heatmap(df[['average_price', 'houses_sold', 'no_of_crimes']].corr(), annot=True, cmap='coolwarm')
plt.title("Correlation Heatmap")
plt.show()
```
